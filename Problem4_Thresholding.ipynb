```python
import numpy as np


# ============================================================
# Problem 4: Change Image Contrast
# ============================================================
# Change image values from one range to another range.


# ============================================================
# Step 1: Inputs and Outputs
# ============================================================
# Input:
# An image/array and two ranges:
# Old range = [old_min, old_max]
# New range = [new_min, new_max]
#
# Output:
# An array containing the values converted to the new range.


# ============================================================
# Step 2: Mathematical Rule
# ============================================================
#
# Formula:
#
# new_value = (value - old_min)
#             / (old_max - old_min)
#             * (new_max - new_min)
#             + new_min
#
# This formula changes values from the old range
# into the new range.


# ============================================================
# Step 3: Manual Trace
# ============================================================

# Example:
#
# Change [50, 100, 150]
# from range [50, 150]
# to range [0, 255].
#
# Formula:
#
# new_value = (value - 50) / (150 - 50) * (255 - 0) + 0
#
# For value = 50:
#
# new_value = (50 - 50) / 100 * 255
#           = 0
#
# For value = 100:
#
# new_value = (100 - 50) / 100 * 255
#           = 50 / 100 * 255
#           = 127.5
#
# For value = 150:
#
# new_value = (150 - 50) / 100 * 255
#           = 255
#
# Therefore:
#
# 50  -> 0
# 100 -> 127.5
# 150 -> 255
#
# Expected output:
# [0, 127.5, 255]


# ============================================================
# Step 4: Python
# ============================================================

def contrast_stretch(
    image: np.ndarray,
    in_min: float,
    in_max: float,
    out_min: float = 0.0,
    out_max: float = 255.0
) -> np.ndarray:
    """Change image values from one range to another."""

    # Convert image values to float
    image_float = image.astype(np.float32)

    # Apply the contrast stretching formula
    stretched = (
        (image_float - in_min)
        / (in_max - in_min)
    )

    stretched = (
        stretched * (out_max - out_min)
        + out_min
    )

    # Keep values inside the output range
    stretched = np.clip(
        stretched,
        out_min,
        out_max
    )

    return stretched


# ============================================================
# Input
# ============================================================

problem_4_input = np.array(
    [50, 100, 150],
    dtype=np.float32
)


# ============================================================
# Calculate Contrast Stretching
# ============================================================

output = contrast_stretch(
    problem_4_input,
    50,
    150
)


# ============================================================
# Display Input and Output
# ============================================================

print("Input Image:")
print(problem_4_input)

print("\nOutput Image:")
print(output)


# ============================================================
# Test
# ============================================================

expected = np.array(
    [0.0, 127.5, 255.0],
    dtype=np.float32
)

np.testing.assert_allclose(
    output,
    expected
)

print("\nProblem 4 passed")
```
